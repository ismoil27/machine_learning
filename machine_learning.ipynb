```python
from sklearn.linear_model import LinearRegression
import pandas as pd
```

```python
data = {
    "area": [50, 80, 100, 120, 150], # feature
    "rooms": [2, 3, 3, 4, 5], # feature
    "age": [20, 15, 10, 5, 2], # feature
    "price": [150, 220, 300, 360, 500] # target
}

df = pd.DataFrame(data)
print(df)
```

```python
X = df[["area", "rooms", "age"]] # feature
y = df["price"] # target
```

# Feature Scaling

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
model = LinearRegression() # step 1

model.fit(X_scaled, y) # step 2
```

```python
new_house = [[110, 3, 7]]

new_house_scaled = scaler.transform(new_house)

pred = model.predict(new_house_scaled)
print('Predicted price:', pred[0])
```

# Underfitting & Overfitting

```python
import numpy as np
import matplotlib.pyplot as plt

# simple test data
X = np.array([1, 2, 3, 4, 5, 6])
y = np.array([1.5, 1.7, 3.2, 3.8, 5.1, 5.5])
```

```python
plt.scatter(X, y)
plt.title('Original Data')
plt.xlabel('X')
plt.ylabel('y')
plt.show()
```

# Underfitting

```python
# from sklearn.linear_model import LinearRegression

model_linear = LinearRegression()
model_linear.fit(X.reshape(-1, 1), y)

y_pred_linear = model_linear.predict(X.reshape(-1, 1))
```

```python
plt.scatter(X, y, label="Real data")
plt.plot(X, y_pred_linear, color="red", label="Linear model")

plt.legend()
plt.title("Underfitting Example")
plt.show()
```

# Overfitting


```python
from sklearn.preprocessing import PolynomialFeatures

poly = PolynomialFeatures(degree=5)
X_poly = poly.fit_transform(X.reshape(-1, 1))

model_poly = LinearRegression()
model_poly.fit(X_poly, y)

y_pred_poly = model_poly.predict(X_poly)
```

```python
plt.scatter(X, y, label="Real data")
plt.plot(X, y_pred_poly, color="green", label="Polynomial (overfit)")

plt.legend()
plt.title("Overfitting Example")
plt.show()
```

```python
plt.scatter(X, y, label="Real data")

plt.plot(X, y_pred_linear, color="red", label="Underfitting")
plt.plot(X, y_pred_poly, color="green", label="Overfitting")

plt.legend()
plt.title("Underfitting vs Overfitting")
plt.show()
```

```python
from sklearn.linear_model import Ridge # Regularization

ridge_model = Ridge(alpha=1.0)
ridge_model.fit(X_poly, y)

y_pred_ridge = ridge_model.predict(X_poly)
```

```python
plt.scatter(X, y, label="Data")

# plt.plot(X, y_pred_poly, label="Overfitting")
plt.plot(X, y_pred_ridge, label="Regularized")

plt.legend()
plt.title("Overfitting vs Regularization")
plt.show()
```
